```python
import pandas as pd
from src.config import DATA_DIR

ligand_df = pd.read_csv(f'{DATA_DIR}/metadata/CROWN_ligand_data.csv')
pdb_df = pd.read_csv(f'{DATA_DIR}/metadata/pdb_metadata.csv')
pli_df = pd.read_csv(f'{DATA_DIR}/metadata/pli_filter_pass.csv')
uniprot_df = pd.read_csv(f'{DATA_DIR}/metadata/cath_full.csv')
special_residues = pd.read_csv(f'{DATA_DIR}/metadata/special_residues.csv')
```

```python
pli_df.rename(columns = {'filename': 'basename'}, inplace = True)
pli_df.head()
```

```python
df = ligand_df.merge(pli_df, how = 'inner', on = ['basename'])
print(len(ligand_df))
print(len(df))
```

```python
df['pdb_id'] = df['basename'].str[:4]
df = df.merge(pdb_df, how = 'inner', on = ['pdb_id'])
print(len(df))
```

```python
df.head()
```

```python
special_residues.rename(columns = {'CCD': 'lig_name'}, inplace = True)
special_residues
```

```python
df_new = df.merge(special_residues, on = 'lig_name', how = 'left', suffixes=("", "_new"))

for col_name in ['MW', 'HeavyAtoms', 'N+O_Atoms', 'HBD', 'HBA', 'RotatableBonds', 'NumRings', 'TPSA', 'QED', 'SMILES', 'MurckoScaffold']:
    df_new[col_name] = df_new[f'{col_name}_new'].combine_first(df_new[col_name])
    df_new.drop(columns = [f'{col_name}_new'], inplace = True)
```

```python
df = df_new.copy()
df.head()
```

```python
# Now merge with UniProt entries
uniprot_df.drop_duplicates(subset = ['pdb_id', 'chain_id'], keep = 'first', inplace = True)
uniprot_df.head()
```

```python
annot_cols = ['uniprot_id', 'taxon_id', 'species_name', 'entry_name',
              'GO_ids', 'protein_name', 'EC_number', 'cath_ids']

multi_cols = ['GO_ids', 'EC_number', 'cath_ids']  # underscore-joined sets

# Pre-split the multi-value columns in df_B into lists
df_B_proc = uniprot_df.copy()
for c in multi_cols:
    df_B_proc[c] = df_B_proc[c].fillna('').astype(str).str.split('_')


# 1. Explode chain_set into one row per (basename, pdb_id, chain_id)
exp = (df[['basename', 'pdb_id', 'chain_set']]
       .assign(chain_id=lambda d: d['chain_set'].str.split('-'))
       .explode('chain_id'))

# 2. Join annotations from df_B on (pdb_id, chain_id)
merged = exp.merge(df_B_proc[['pdb_id', 'chain_id', *annot_cols]],
                   on=['pdb_id', 'chain_id'], how='left')

# 3. Collapse to one row per basename, collecting unique values as sets
def to_set(s):
    out = set()
    for v in s.dropna():
        if isinstance(v, list):
            out.update(x for x in v if x != '')
        else:
            out.add(v)
    return out

agg = merged.groupby('basename')[annot_cols].agg(to_set)

# 4. Attach back to df_A
df = df.merge(agg, on='basename', how='left')
```

```python
df.head()
```

```python
set_cols = ['uniprot_id', 'taxon_id', 'species_name', 'entry_name',
            'GO_ids', 'protein_name', 'EC_number', 'cath_ids']

df_out = df.copy()
for c in set_cols:
    df_out[c] = df_out[c].apply(lambda x: sorted(x) if isinstance(x, set) else x)

df_out.to_parquet('annotated.parquet')
```
